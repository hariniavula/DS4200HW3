```python
import pandas as pd

df = pd.read_csv('socialMedia.csv')

avg_data = df.groupby(['Platform', 'PostType'])['Likes'].mean().reset_index()
avg_data.columns = ['Platform', 'PostType', 'AvgLikes']
avg_data['AvgLikes'] = avg_data['AvgLikes'].round(2)

avg_data.to_csv('socialMediaAvg.csv', index=False)

print(avg_data)
```

```python
import pandas as pd

# Read the original CSV file
df = pd.read_csv('socialMedia.csv')


avg_time_data = df.groupby('Date')['Likes'].mean().reset_index()
avg_time_data.columns = ['Date', 'AvgLikes']

avg_time_data['DateOnly'] = pd.to_datetime(avg_time_data['Date'].str.split(' ').str[0], format='%m/%d/%Y')
avg_time_data = avg_time_data.sort_values('DateOnly')
avg_time_data = avg_time_data.drop('DateOnly', axis=1)

avg_time_data = avg_time_data.reset_index(drop=True)

avg_time_data.to_csv('socialMediaTime.csv', index=False)

print(avg_time_data)
```
